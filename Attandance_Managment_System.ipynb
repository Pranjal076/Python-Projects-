```python
import numpy as np
import pandas as pd

# Step 1: Create a list of students
students = ["Pranjal", "Harshad", "Shreeansh", "Tarang", "Gargi", "Pranjal_J"]

# Step 2: Initialize attendance DataFrame
# Let's assume 7 days of classes
days = ["Day1", "Day2", "Day3", "Day4", "Day5", "Day 6", "Day 7"]

# Random attendance: 1 = Present, 0 = Absent
#attendance_data = np.random.randint(0, 2, size=(len(students), len(days)))
attendance_data = [
    [1, 1, 1, 1, 0, 1, 0],  # Pranjal
    [1, 1, 1, 0, 0, 0, 1],  # Harshad
    [0, 1, 1, 1, 1, 1, 1],  # Shreeansh
    [1, 1, 1, 1, 0, 1, 1],  # Tarang
    [0, 0, 1, 0, 1, 0, 0],  # Gargi
    [1, 0, 0, 0, 0, 1, 0]   # Pranjal_J
]

#df = pd.DataFrame(attendance_data, index=students, columns=days)


df = pd.DataFrame(attendance_data, index=students, columns=days)

print("Attendance Sheet:\n")
print(df)

# Step 3: Calculate total attendance per student
df["Total_Present"] = df.sum(axis=1)

# Step 4: Calculate attendance percentage
df["Attendance_%"] = (df["Total_Present"] / len(days)) * 100

print("\nFinal Attendance Report:\n")
print(df)

# Step 5: Mark students with < 75% attendance
df["Status"] = np.where(df["Attendance_%"] >= 75, "Eligible", "Not Eligible")

print("\nEligibility Report:\n")
print(df[["Attendance_%", "Status"]])
```

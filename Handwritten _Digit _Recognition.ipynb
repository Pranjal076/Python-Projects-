```python
import tensorflow as tf
from tensorflow. keras import Sequential, Input, layers

# Step 1: Load MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

# Step 2: Normalize pixel values (0–255 → 0–1)
# xnormalized = x/255
X_train, X_test = X_train / 255.0, X_test / 255.0

# Step 3: Build the model
model = Sequential([
    Input(shape=(28, 28)),          # ✅ Correct way to define input
    layers.Flatten(),               # Flatten 28x28 images into 784 vector
    layers.Dense(128, activation='relu'),  # Hidden layer
    layers.Dense(10, activation='softmax') # Output layer (10 digits)
])

# Step 4: Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Step 5: Train the model
model.fit(X_train, y_train, epochs=6, validation_data=(X_test, y_test))

# Step 6: Evaluate the model
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)
print("\nTest accuracy:", test_acc)
```
